##4. Class Variables and Class Methods

```python
class Bank:
  bank_name = "Anonymous"

  @classmethod
  def change_bank_name(cls, name):
    cls.bank_name = name

bank1 = Bank()
bank2 = Bank()
print(f"Bank1: {bank1.bank_name}")
print(f"Bank2: {bank2.bank_name}")

Bank.change_bank_name("UBL")
print(f"\nBank1: {bank1.bank_name}")
print(f"Bank2: {bank2.bank_name}")

bank3 = Bank()
print(f"\nBank3: {bank3.bank_name}")
```

##8. The super() Function

```python
class Person:
  def __init__(self, name):
    self.name = name

class Teacher(Person):
  def __init__(self, name, subject):
    self.subject = subject
    super().__init__(name)

  def teacher_details(self):
    print("Teacher Details")
    print(f"Name: {self.name}")
    print(f"Subject: {self.subject}")

teacher = Teacher("Ali", "Typescript")
teacher.teacher_details()
```

##9. Abstract Classes and Methods

```python
from abc import ABC, abstractmethod
class Shape(ABC):

  @abstractmethod
  def area(self):
    print("Area of Shape")

class Rectangle(Shape):

  def area(self, length, width):
    area = length * width
    print(f"Area of rectangle is: {area}")

rect = Rectangle()
area_of_rect = rect.area(5, 4)
```

##14. Aggregation

```python
class Employee:
  def __init__(self, name, post):
    self.name = name
    self.course = post

  def __str__(self):
    return f"{self.name} work as {self.course}"

class Department:
  def __init__(self, dept):
    self.dept = dept
    self.employees = []

  def add_emp(self, emp):
    self.employees.append(emp)

  def display_emp(self):
    print(f"Employees in {self.dept}")
    for emp in self.employees:
      print(f"- {emp}")

emp1 = Employee("Muhammad Rayyan Naveed", "Frontend Developer")
emp2 = Employee("Syed Zohaib Shah", "Data Analyst")

cs_dept = Department("IT")

cs_dept.add_emp(emp1)
cs_dept.add_emp(emp2)

cs_dept.display_emp()
```

##15. Method Resolution Order (MRO) and Diamond Inheritance

```python
from ast import Pass
class A:
  def show(self):
    return "Show A"

class B(A):
  def show(self):
    return "Show B"

class C(A):
  def show(self):
    return "Show C"

class D(B, C):
    pass

d = D()
print(D.mro())
d.show()
```

##16. Function Decorators

```python
def log_function_call(func):
  def wrapper():
    print("Function is being called")
    return func()
  return wrapper

@log_function_call
def say_hello():
  print("Hi Muhammad Rayyan Naveed")

say_hello()
```

##17. Class Decorators

```python
def add_greetings(cls):
  def greet(self):
    return "Hello From Decorator!"

  cls.greet = greet
  return cls

@add_greetings
class Person:
  pass

person = Person()
print(person.greet())
```

##18. Property Decorators: @property, @setter, and @deleter

```python
class Product:
  def __init__(self, price):
    self.__price = price

  @property
  def price(self):
    return self.__price

  @price.setter
  def price(self, new_price):
    self.__price = new_price

  @price.deleter
  def price(self):
    del self.__price

product = Product(10)
product.price
product.price = 20
product.price
# del product.price
# product.set_price(50)
# product.price
```

##19. callable() and __call__()

```python
class Multiplier:
  def __init__(self, factor):
    self.factor = factor

  def __call__(self, value):
    return value * self.factor

double = Multiplier(2)
print(f"Is double callable {callable(double)}")

calling_obj = double(20)
print(f"Is object callable: {calling_obj}")
```

##20. Creating a Custom Exception

```python
class InvalidAgeError(Exception):
  """Exception raised when age is less than 18."""
  pass

def check_age(age):
  if age < 18:
    raise InvalidAgeError("Age must be at least 18.")
  else:
    print("You are adult")


try:
  user_age = int(input("Enter your age: "))
  check_age(user_age)

except InvalidAgeError as e:
  print("InvalidAgeError:", e)

except ValueError:
  print("Please enter a valid integer for age.")
```

##21. Make a Custom Class Iterable

```python
class Countdown:
  def __init__(self, start_num):
    self.current= start_num

  def __iter__(self):
    return self

  def __next__(self):
    if self.current < 0:
      raise StopIteration

    value = self.current
    self.current -= 1
    return value

for num in Countdown(10):
  print(num)
```
