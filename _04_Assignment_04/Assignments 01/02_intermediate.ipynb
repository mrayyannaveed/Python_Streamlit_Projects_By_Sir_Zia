```python
import random

num_rounds = 5

def main():
  print("Welcome to the High Low game")
  print("----------------------------")

  your_score = 0
  for i in range(num_rounds):
    print(f"Rounds: {i + 1}")
    computer_num = random.randint(1, 100)
    user_num = int(input("Your number is: "))

    choice = input("Do you think your number is higher or lower than the computer's: ")

    while choice != "higher" and choice != "lower":
      choice = input("Please enter either higher or lower: ")

    higher_and_correct = choice == "higher" and user_num > computer_num
    lower_and_correct = choice == "lower" and user_num < computer_num

    if higher_and_correct or lower_and_correct:
      print(f"You were right! The computer number was: {computer_num}")
      your_score += 1
    else:
      print(f"That's incorrect the computer number was: {computer_num}")

    # print(f"Your score is now: {your_score}")
    print()

  print(f"Your final score is: {your_score}")
  if your_score == num_rounds:
    print("Wow! You played perfectly")
  elif your_score > num_rounds // 2:
    print("Good Job! You played really well!")
  else:
    print("Better Luck Next Time")

  print("Thanks for playing")

if __name__ == "__main__":
  main()
```

```python
Mercury_Gravity = 0.376
Venus_Gravity = 0.889
Mars_Gravity = 0.378
Jupiter_Gravity = 2.36
Saturn_Gravity = 1.081
Uranus_Gravity = 0.815
Neptune_Gravity = 1.14

def main():
  weight_on_earth = float(input("Enter your weight on  earth: "))
  planet = input("Enter a planet: ")

  if planet == "Mercury":
    gravity_constant = Mercury_Gravity
  elif planet == "Venus":
    gravity_constant = Venus_Gravity
  elif planet == "Mars":
    gravity_constant = Mars_Gravity
  elif planet == "Jupiter":
    gravity_constant = Jupiter_Gravity
  elif planet == "Saturn":
    gravity_constant = Saturn_Gravity
  elif planet == "Uranus":
    gravity_constant = Uranus_Gravity
  else:
    gravity_constant = Neptune_Gravity

  weight_on_planet = weight_on_earth * gravity_constant
  rounded_planet_weight = round(weight_on_planet, 2)

  print(f"The equivalent weight on {planet}: {str(rounded_planet_weight)}")


if __name__ == "__main__":
  main()
```

```python
def access_element(lst, index):
  try:
    return lst[index]
  except IndexError:
    return "Index is out of range"


def modify_element(lst, index, new_element):
  try:
    lst[index] = new_element
    return lst
  except IndexError:
    return "Index is out of range"

def slice_elements(lst, start, end):
  try:
    return lst[start:end]
  except IndexError:
    return "Invalid Indices"


def index_game():
  operation = input("Select an operation (access, modify, slice): ")
  elements = ["cultus", 8, "corolla", 3, "civic", 23]

  if operation == "access":
    index = int(input("Enter index: "))
    elements = access_element(elements, index)
    print(elements)

  elif operation == "modify":
    value = input("Enter element to modify: ")
    index = int(input("Enter index: "))
    elements = modify_element(elements, index, value)
    print(elements)

  elif operation == "slice":
    start = int(input("Start index: "))
    end = int(input("End index: "))
    elements = slice_elements(elements, start, end)
    print(elements)

  else:
    print("Invalid Operation")

index_game()
```
